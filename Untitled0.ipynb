# **Gráficos estadísticos**






A continuación se visualizará la creación e interpretación de los gráficos estadísticos respecto a las variables de estudio, iniciamos con el histograma:


```python
import matplotlib.pyplot as plot
intervalos = range(min(df.EDAD_PADRE), max(df.EDAD_PADRE) + 1, 5)
plot.hist(df.EDAD_PADRE, bins=intervalos, color='darkgreen', rwidth=0.85)
plot.title('Histograma de la edad de un padre')
plot.xlabel('Edad (años)')
plot.ylabel('Frecuencia')
plot.xticks(intervalos)
plot.show()
```

En el histograma de la variable "Edad de un padre" se puede extraer la información de que la frecuencia está sesgada por derecha, además de tener múltiples picos.

Por otra parte, se pueden observar valores atípicos muy mínimos, para luego no obtener más datos respecto a la edad del padre; En cambio hay una concentración de edades en un rango específico, donde este se podría indicar como preferencias o tendencias en la edad de los padres en el año 2020.

Podemos concluir que con el histograma no podemos obtener datos especificos, ya que contiene valores muy grandes, o sea, son valores por encima de 100.

```python
import matplotlib.pyplot as plot
intervalos = range(min(df.N_EMB), max(df.N_EMB) + 1, 5)
plot.hist(df.N_EMB, bins=intervalos, color='darkblue', rwidth=0.85)
plot.title('Histograma del número de embarazos')
plot.xlabel('Número de embarazo')
plot.ylabel('Frecuencia')
plot.xticks(intervalos)
plot.show()
```

Con base al histograma anterior, podemos observar que gráficamente hay un sesgo por derecha de las personas encuestadas en lo cual no presentan más de 16 embarazos, tomando esto como un valor atípico desde los 16 embarazos hasta los 99.

En cambio, se logra visualizar un pico en específico de número de embarazos que va desde tener 1 hasta 6 embarazos, los demás picos significan que son números de embarazos no tan comunes en que lleguen a ocurrir.

Esto nos quiere decir que, entre las personas encuentadas no existe ninguna que hubiese tenido más de 16 embarazos y la cantidad de embarazos más común se encuentra entre entre 1 y 6 embarazos; sin embargo, existe personas que han tenido entre 11 y 16 embarazos.

```python
df1 = df[(df.EDAD_PADRE < 120) & (df.N_EMB != 99) & (df.N_HIJOSV != 99)]
df1.head(5)
```

```python
intervalos1 = range(min(df1.EDAD_PADRE), max(df1.EDAD_PADRE) + 1)
plot.figure(figsize=(10, 6))
sns.histplot(x = df1.EDAD_PADRE, bins = intervalos1, kde = True, stat = "density", color = "skyblue", edgecolor='black') # kde = True es la gráfica de la función de densidad de probabilidad estimada por un método Kernell

xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 1000)
p = norm.pdf(x, np.mean(df1.EDAD_PADRE), np.std(df1.EDAD_PADRE))

plot.plot(x, p, 'k', linewidth=2, label='Distribución Normal', color = "darkblue")
sns.kdeplot(x=df1.EDAD_PADRE, color = "darkgreen")

plot.title('Histograma, FDP por un Kernel y FDP bajo Normalidad')
plot.xlabel('Edad (años)')
plot.ylabel('Densidad')
plot.legend()

# Mostrar el gráfico
plt.show()
```

Sea $X:=$ "Edad del padre de un recien nacido en colombia"

Descriptivamente el histograma y el fdp de Kernel no coninciden totalmente, significa que la distribución de edades de los padres presenta asimetrías. Así mismo, esto indica que como la función de densidad de probabilidad de los datos estimada por un método Kernel (línea verde) no coincide con la función de densidad de probabilidad bajo normalidad (línea azul), es decir, $x\sim\text{Normal}(\text{Media}=\bar{x}, \text{Varianza}= s_x)$, entonces la variable edad del padre de un recien nacido en Colombia no asume normalidad.

```python
intervalos1 = range(min(df1.N_EMB), max(df1.N_EMB) + 1)
plot.figure(figsize=(10, 6))
sns.histplot(x = df1.N_EMB, bins = intervalos1, kde = True, stat = "density", color = "green", edgecolor='black') # kde = True es la gráfica de la función de densidad de probabilidad estimada por un método Kernell

xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 1000)
p = norm.pdf(x, np.mean(df1.N_EMB), np.std(df1.N_EMB))

plot.plot(x, p, 'k', linewidth=2, label='Distribución Normal', color = "purple")
sns.kdeplot(x=df1.N_EMB, color = "blue")

plot.title('Histograma, FDP por un Kernel y FDP bajo Normalidad')
plot.xlabel('Edad (años)')
plot.ylabel('Densidad')
plot.legend()

# Mostrar el gráfico
plt.show()
```

Sea $Y:=$ "El número de embarazos incluido el presente"

Descriptivamente el histograma y el fdp de Kernel no coninciden totalmente, esto significa que la distribución de el número de embarazos presenta asimetrías.

Como la función de densidad de probabilidad de los datos estimada por un método Kernel (línea azul) no coincide con la función de densidad de probabilidad bajo normalidad (línea morada), es decir, $x\sim\text{Normal}(\text{Media}=\bar{x}, \text{Varianza}= s_x)$, entonces no asumimos normalidad para la variable $Y=$"Número de embarazos inluidos el presente"; esto deacuerdo con las gráficas antes presentadas

```python
shapiro(df1.EDAD_PADRE) #Prueba de hipótesis
```

**La hipótesis nula** = La varibale aleatoria $X$ presenta distribución normal

*  Si el **valor p** en una prueba de hipótesis es menor que 0.05 se rechaza la hipótesis nula con una confianza del 95%

* Si el **valor p** en una prueba de hipótesis es mayor que 0.05  NO se rechaza la hipótesis nula con una confianza del 95%

En nuestro caso particular, tenemos que pvalue=$0.0$ < 0.05, es decir, se rechaza que la variable aleatoria X, es decir, la edad de un padre de un recién nacido en Colombia, presenta distribución normal y este rechazo se hace con un 95% de confianza.

```python
shapiro(df1.N_EMB) #Prueba de hipótesis
```

**La hipótesis nula** = La variable aleatoria $Y$ presenta distribución normal

*   Si el valor p en una prueba de hipótesis es mayor que 0.05 NO se rechaza la hipótesis nula con una confianza del 95%

Por medio del valor anterior obtenemos que pvalue=$0.0$ < 0.05, es decir, se rechaza con un 95% de confianza la hipótesis nula, lo cual es que el número de embarazos incluido el presente presenta distrbución normal.

En esta parte del proyecto, analizaremos la correlación entre la variable "EDAD_PADRE" y la variable "N_EMB". Nuestro objetivo es determinar la
existencia de una relación entre estos conjuntos de datos bivariados y calcular el grado de correlación correspondiente.

```python
io_csv = io.BytesIO(uploaded['asriosgu_nac2020.csv'])
io_csv = io.BytesIO(uploaded['asriosgu_nac2020.csv'])
tabla = pd.read_csv(io_csv, sep = ",", index_col=0, encoding='latin-1')
tabla.head(3)
```

```python
import pandas as pd
import io
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy.stats import norm
import matplotlib.pyplot as plot
from scipy.stats import shapiro
import plotly.express as px
from scipy.stats import skew, kurtosis
from sklearn.linear_model import LinearRegression
from scipy.stats import spearmanr, kendalltau
import plotly.express as px
```

```python
df=tabla[['EDAD_PADRE', 'N_EMB']]
df1 = df[(df["N_EMB"] < 40) & (df["EDAD_PADRE"]  < 120)]
df1.head(3)
```

# **Gráficos Estadísticos**

```python

plt.figure(figsize=(5,5))
plt.scatter(df1["EDAD_PADRE"], df1["N_EMB"], c="darkblue", marker="o")
plot.title('Diagrama de Dispersión')
plot.xlabel("Edad del padre")
plot.ylabel("Número de embarazos")
plt.show()


```

Del diagrama presentado se observa una dispersión de puntos sin un patrón particular, de donde se infiere que las variables "EDAD_PADRE" y "N_EMB" son independientes y no están correlacionadas.
Es decir, en el diagrama de dispersión nos muestra que la edad del padre no influye en la frecuencia de embarazos en las mujeres. No podemos afirmar que a medida que aumenta la edad del padre el número de embarazos de las mujeres colombianas varíe significativamente.

```python
modelo = LinearRegression()
modelo.fit(np.array(df1["EDAD_PADRE"]).reshape(-1,1), np.array(df1['N_EMB']))
predi_lineal = modelo.predict((np.array(df1["EDAD_PADRE"]).reshape(-1,1)))
predi_lineal
```

```python
plt.figure(figsize=(5,5))
plt.scatter(df1["EDAD_PADRE"], df1["N_EMB"], c="pink", marker="o")
plt.plot(df1["EDAD_PADRE"], predi_lineal, linestyle='-', label= "OLS")
plot.title('Diagrama de Dispersión')
plot.xlabel("Edad del padre")
plot.ylabel("Número de embarazos")
plt.show()
```

De acuerdo con la recta de regresión se determina una correlación postiva, es decir se observa que entre más años tiene el padre el número de embarazos en las mujeres colombianas tiende a aumentar.

```python
coef_pearson = np.corrcoef(df1["EDAD_PADRE"], df1["N_EMB"])[0,1]
coef_spearman, _= spearmanr(df1["EDAD_PADRE"], df1["N_EMB"])
coef_kendall, _= kendalltau(df1["EDAD_PADRE"], df1["N_EMB"])
print(f"Los coeficientes de relación de Pearson, Spearman, spearman y Kendall son respectivamente {coef_pearson}, {coef_spearman} y {coef_kendall}.")
```

Para la interpretación de los coeficientes de correlación nos guiaremos de la siguiente referencia: Landis, J. R., & Koch, G. G. (1977). The measurement of observer agreement for categorical data. biometrics, 159-174.[Link](https://pdfs.semanticscholar.org/7e73/43a5608fff1c68c5259db0c77b9193f1546d.pdf)

```python
pearson_corr = np.corrcoef(df['N_EMB'], df['EDAD_PADRE'])[0,1]
pearman_corr,_=spearmanr(df['N_EMB'], df['EDAD_PADRE'])
kendall_corr_=kendalltau(df['N_EMB'], df['EDAD_PADRE'])
```

# **Conclusiones**

Después de analizar las variables **"Edad del padre de un recién nacido en Colombia" (X) y "Número de embarazos, incluyendo el actual" (Y)**, se puede concluir lo siguiente:

**1.** Las variables X e Y no siguen una distribución normal, tal como lo indican los resultados de las pruebas de normalidad realizadas. Esto se evidencia en las asimetrías observadas en los histogramas y en la falta de ajuste entre las funciones de densidad de probabilidad estimadas y las funciones de densidad bajo normalidad.

**2.** La variable X presenta una distribución sesgada positivamente, con múltiples picos en el histograma, lo que sugiere preferencias o tendencias en la edad de los padres de recién nacidos en Colombia durante el año 2020. Además, se observa una alta variabilidad en las edades, con un rango intercuartílico de 10 años.

**3.** La variable Y, correspondiente al número de embarazos, también muestra una distribución asimétrica, pero en este caso es una asimetría negativa, indicando que la media es menor que la moda. La variable Y presenta una distribución leptocúrtica, es decir, con una alta concentración de valores alrededor de la media.

Por último, respecto a la correlación entre estas variables y el grado de correlación, las variables "EDAD_PADRE" y "N_EMB" están discretamente correlacionadas. Puesto que existe una correlación positiva débil, que nos fue mostrada por la recta de regresión, esto significa que hay una leve tendencia a que el número de embarazos aumente con la edad del padre. Sin embargo, esta relación no es lo suficientemente fuerte como para que nos permita afirma que la edad del padre sea un factor determinante en el número de embarazos de las mujeres colombianas. La variabilidad en el número de embarazos no está significativamente explicada por la edad del padre, según los datos analizados.
